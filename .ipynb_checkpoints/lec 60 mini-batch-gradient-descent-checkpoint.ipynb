```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X, y = load_diabetes(return_X_y = True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
y_pred = reg.predict(X_test)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
r2_score(y_test, y_pred)
```

# Our's Mini-batch GD regressor

```python
import random
class MGDRegressor: 

    def __init__(self,batch_size, learning_rate = 0, epochs = 100):
        self.intercept_= None
        self.coef_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size

    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        for i in range(int(X_train.shape[0]/self.batch_size)):
            # update all coef and intercept
            # idx = random.sample(range(X_train.shape[0], self.batch_size)) this was my mistake
            
            idx = random.sample(range(X_train.shape[0]),self.batch_size)
            y_hat = np.dot(X_train[idx],self.coef_)+self.intercept_
            intercept_der = -2*np.mean(y_train[idx] - y_hat)
            self.intercept_ = self.intercept_ - (self.lr*intercept_der)

            coef_der = -2*np.dot((y_train[idx] - y_hat), X_train[idx])
            self.coef_ = self.coef_ -(self.lr*coef_der)
            
        print(self.coef_, self.intercept_)

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
X_test.shape
```

```python
mgdr = MGDRegressor(batch_size=int(X_train.shape[0]/60), learning_rate = 0.5, epochs = 100)
```

```python
mgdr.fit(X_train, y_train)
```

```python
y_pred = mgdr.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

# SKlearn Mini-batch GD

```python
from sklearn.linear_model import SGDRegressor
```

```python

sgd = SGDRegressor(learning_rate='constant',eta0=0.1)
```

```python
batch_size = 35

for i in range(200):
    
    idx = random.sample(range(X_train.shape[0]),batch_size)
    sgd.partial_fit(X_train[idx],y_train[idx])
```

```python
sgd.coef_
```

```python
sgd.intercept_
```

```python
y_pred = sgd.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
