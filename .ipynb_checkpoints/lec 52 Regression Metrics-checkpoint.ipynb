```python
plt.figure(figsize=(8, 5))
plt.scatter(df['age'], df['charges'], color='blue', alpha=0.5)
plt.title('Age vs Medical Charges')
plt.xlabel('Age (Years)')
plt.ylabel('Charges ($)')
plt.grid(True)
plt.show()
```

```python
smokers = df[df['smoker'] == 'yes']
non_smokers = df[df['smoker'] == 'no']

plt.figure(figsize=(8, 5))
plt.scatter(non_smokers['age'], non_smokers['charges'], color='green', label='Non-Smoker', alpha=0.6)
plt.scatter(smokers['age'], smokers['charges'], color='red', label='Smoker', alpha=0.6)
plt.title('Age vs Charges by Smoking Status')
plt.xlabel('Age')
plt.ylabel('Charges ($)')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.hist(df['charges'], bins=30, color='purple', edgecolor='black')
plt.title('Distribution of Medical Charges')
plt.xlabel('Charges ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
# convert text cols into dummy numbers 0 and 1
df_encoded = pd.get_dummies(df, columns=['sex', 'smoker', 'region'], drop_first=True)

# seperate x and y
X = df_encoded.drop(columns=['charges'])
y = df_encoded['charges']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.20, random_state = 2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
print("MAE: ", mae)
```

```python
mse = mean_squared_error(y_test, y_pred)
print("MAE: ", mse)
```

```python
rmse = np.sqrt(mse)
print("RMSE: ", rmse)
```

```python
r2 = r2_score(y_test, y_pred)
print("R2_Score: ", r2)
```

```python
# Adjusted R2 Score

n = X_test.shape[0]
k = X_test.shape[1]
```

```python
adj_r2 = 1 - ((1 - r2) * (n - 1) / (n - k - 1))
print("Adjusted R2 Score: ", adj_r2)
```

```python
# Mean Absolute Error (MAE = ~4,279.86): On average, the model's predicted insurance charges deviate from the actual charges by approximately $4,280.Mean Squared Error (MSE = ~38,304,871.35): Represents the squared error penalty. The value is large because it squares the dollar differences ($\$^{2}$), severely penalizing larger prediction errors.Root Mean Squared Error (RMSE = ~6,189.09): Converts the squared error back to dollars. Because RMSE ($6,189) is notably higher than MAE ($4,280), it confirms the presence of significant outlier errors (e.g., extreme medical charges from smokers) where the model made large mispredictions.$R^2$ Score (0.7449): Approximately 74.5% of the variance in insurance charges is explained by the input features using this Linear Regression model.Adjusted $R^2$ Score (0.7371): Slightly lower than $R^2$ at 73.7%, accounting for the total number of predictors and penalizing redundant features. The small difference indicates that most included features contribute meaningful explanatory power.
```
